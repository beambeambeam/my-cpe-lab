## Cancer Poisson-regression interpretation

The model estimates the expected number of deaths from age group, smoking category, and population. Poisson coefficients are log-count effects, so positive coefficients indicate a higher expected death count than the reference group, holding the other predictors constant.

All fitted age-group, smoking-category, and population coefficients are statistically significant in the model summary (p < 0.001). The age coefficients are generally larger for older groups, while the smoking and population coefficients are also positive. Predictors were min–max scaled, so each coefficient describes a change across that predictor's observed range. These are associations in this dataset, not proof of causation.

# 8. Survival analysis

For this example, we will be investigating the lifetimes of political leaders around the world. A political leader, in this case, is defined by a single individual’s time in office who controls the ruling regime. This political leader could be an elected president, unelected dictator, monarch, etc. The birth event is the start of the individual’s tenure, and the death event is the retirement of the individual. Censoring can occur if they are a) still in offices at the time of dataset compilation (2008), or b) die while in power (this includes assassinations).

For example, the Bush regime began in 2000 and officially ended in 2008 upon his retirement, thus this regime’s lifespan was eight years, and there was a “death” event observed. On the other hand, the JFK regime lasted 2 years, from 1961 and 1963, and the regime’s official death event was not observed – JFK died before his official retirement.

# Mathematical Notations

## Survival function

We will assume for now that T is a continuous random variable with probability density function (p.d.f.) $$f(t)$$ and cumulative distribution function (c.d.f.) $$F(t)=Pr(T<t),$$ giving the probability that the event has occurred by duration t.

$$S(t) = \Pr\{ T \ge t \} = 1 - F(t) =  \int_t^\infty f(x) dx,$$

## Hazard function

An alternative characterization of the distribution of T
 is given by the hazard function, or instantaneous rate of occurrence of the event, defined as

$$\lambda(t) = \lim_{dt\rightarrow0} \frac{\Pr\{t \le T < t + dt |
T \ge t \} }{dt}.$$

# 9. Cox’s proportional hazard model

The idea behind Cox’s proportional hazard model model is that the log-hazard of an individual is a linear function of their static covariates and a population-level baseline hazard that changes over time. Mathematically:

$$\underbrace{h(t | x)}_{\text{hazard}} = \overbrace{b_0(t)}^{\text{baseline hazard}} \underbrace{\exp \overbrace{\left(\sum_{i=1}^n b_i (x_i - \overline{x_i})\right)}^{\text{log-partial hazard}}}_ {\text{partial hazard}}$$

```python
from lifelines import CoxPHFitter
from lifelines.datasets import load_rossi

rossi_dataset = load_rossi()
rossi_dataset.head()
```

This data set is originally from Rossi et al. (1980), and is used as an example in Allison (1995). The data pertain to 432 convicts who were released from Maryland state prisons in the 1970s and who were followed up for one year after release. Half the released convicts were assigned at random to an experimental treatment in which they were given financial aid; half did not receive aid.

- week: week of first arrest after release or censoring; all censored observations are censored at 52 weeks.
- arrest: 1 if arrested, 0 if not arrested.
- fin: financial aid: no yes.
- age: in years at time of release.
- race: black or other.
- wexp: full-time work experience before incarceration: no or yes.
- mar: marital status at time of release: married or not married.
- paro: released on parole? no or yes.
- prio: number of convictions prior to current incarceration.

# Activity - Estimating Lifetime
- Use Telco Churn dataset and build a Kaplan Meier survival analysis.
- Compare lifetime between different payment methods.

We treat customer churn as the event. Customers who have not churned are right-censored at their observed tenure.

```python
import pandas as pd
import matplotlib.pyplot as plt
from lifelines import KaplanMeierFitter

telco = pd.read_csv("Telco-Churn.csv")
telco["tenure"] = pd.to_numeric(telco["tenure"], errors="coerce")
telco["churned"] = telco["Churn"].eq("Yes").astype(int)
telco = telco.dropna(subset=["tenure", "PaymentMethod"]).copy()

telco[["tenure", "PaymentMethod", "Churn", "churned"]].head()
```

## Fit and plot Kaplan–Meier curves

Each curve estimates the probability that a customer remains active after a given number of months.

```python
fig, ax = plt.subplots(figsize=(10, 6))
km_fitters = {}

for payment_method, group in telco.groupby("PaymentMethod", sort=True):
    kmf_telco = KaplanMeierFitter(label=payment_method)
    kmf_telco.fit(
        durations=group["tenure"],
        event_observed=group["churned"],
    )
    kmf_telco.plot_survival_function(ax=ax)
    km_fitters[payment_method] = kmf_telco

ax.set(
    title="Telco customer lifetime by payment method",
    xlabel="Tenure (months)",
    ylabel="Probability of remaining",
    ylim=(0, 1),
)
ax.grid(alpha=0.25)
plt.tight_layout()
plt.show()
```

## Compare estimated lifetime

The table reports sample size, churn events, median estimated lifetime, and survival at 12 and 24 months.

```python
comparison = []

for payment_method, group in telco.groupby("PaymentMethod", sort=True):
    kmf_telco = km_fitters[payment_method]
    comparison.append(
        {
            "Payment method": payment_method,
            "Customers": len(group),
            "Churned": int(group["churned"].sum()),
            "Median lifetime (months)": kmf_telco.median_survival_time_,
            "Survival at 12 months": float(
                kmf_telco.survival_function_at_times(12).iloc[0]
            ),
            "Survival at 24 months": float(
                kmf_telco.survival_function_at_times(24).iloc[0]
            ),
        }
    )

comparison = pd.DataFrame(comparison).sort_values("Median lifetime (months)")
display(DualFormatTable(comparison))
```

## Test whether payment-method curves differ

A multivariate log-rank test compares all payment-method survival curves jointly.

```python
from lifelines.statistics import multivariate_logrank_test

logrank = multivariate_logrank_test(
    event_durations=telco["tenure"],
    groups=telco["PaymentMethod"],
    event_observed=telco["churned"],
)
logrank_table = logrank.summary.copy()
logrank_table["p"] = logrank_table["p"].map(lambda value: f"{value:.3e}")
display(DualFormatTable(logrank_table))
```

## Telco survival-analysis conclusion

Electronic-check customers have the shortest estimated median lifetime: 47 months. The bank-transfer, credit-card, and mailed-check curves do not reach 0.5 survival during the observed 0–72 month window, so their median lifetime is not reached in this dataset; inf does not mean infinite real-world lifetime.

At 24 months, estimated survival is about 62.4% for electronic check, 90.2% for bank transfer, 91.3% for credit card, and 79.7% for mailed check. The multivariate log-rank test gives chi-square(3) = 865.24 and p = 3.07 x 10^-187, so we reject the hypothesis that all payment-method survival curves are equal. This shows an association between payment method and observed customer lifetime, not causation.
